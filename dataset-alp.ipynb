```python
cosine_1 = np.cos(np.linspace(0, 2 * np.pi, 1000))
cosine_2 = np.concatenate((cosine_1[500:], cosine_1[:500]))
cosine_3 = np.concatenate((cosine_1[250:], cosine_1[:250]))

plt.plot(cosine_1, label='cosine wave')
plt.plot(cosine_2, label='shifted cosine wave')
plt.plot(cosine_3, label='another shifted cosine wave')
```

```python
for x in np.linspace(0, 1, 11):
    plt.plot(cosine_3*x, alpha=0.2, color='red')

plt.show()
```

```python
for x in np.linspace(-1, 1, 11):
    plt.plot((cosine_3+x)/2, alpha=0.2, color='red')
    
plt.show()
```
